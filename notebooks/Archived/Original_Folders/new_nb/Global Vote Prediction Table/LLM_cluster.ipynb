```python
import os

# Path to the folder containing the classified datasets
classified_folder = os.path.join("..", "..", "..", "data", "classified_llm_datasets")

# Initialize an empty list to store DataFrames
classified_dfs = []

# There are 88 files, named classified_voting_texts_batch_{index}.csv
for idx in range(1, 89):
    file_path = os.path.join(classified_folder, f"classified_voting_texts_batch_{idx}.csv")
    if os.path.exists(file_path):
        temp_df = pd.read_csv(file_path, dtype={'idVotacao': str})
        classified_dfs.append(temp_df[['idVotacao', 'classification']])
    else:
        print(f"Warning: {file_path} does not exist.")

# Concatenate all classified DataFrames if any were loaded
if classified_dfs:
    classified_all = pd.concat(classified_dfs, ignore_index=True)

    # Ensure the id columns are strings for matching
    df['id'] = df['id'].astype(str)
    classified_all['idVotacao'] = classified_all['idVotacao'].astype(str)

    # Merge the classification into the main df
    df = df.merge(classified_all, how='left', left_on='id', right_on='idVotacao')

    # Rename the column to classified_llm and drop the helper column
    df.rename(columns={'classification': 'classified_llm'}, inplace=True)
    df.drop(columns=['idVotacao'], inplace=True)
else:
    print("No classified datasets were loaded. The 'classified_llm' column will not be added.")

df.tail(5)
```

```python
nan_count = df['classified_llm'].isna().sum()
not_nan_count = df['classified_llm'].notna().sum()
print(f"NaN count in 'classified_llm': {nan_count}")
print(f"Not NaN count in 'classified_llm': {not_nan_count}")

```
