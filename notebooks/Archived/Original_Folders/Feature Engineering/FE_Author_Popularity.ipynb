```python
import pandas as pd
```

```python
df_sessions = pd.read_csv("df_all_info.csv")
```

```python
df_sessions['id'].duplicated().sum()
```

```python
df_sessions_unique = df_sessions.drop_duplicates(subset=['id'])
```

```python
# Define the path pattern and years
data_path = "../data/voting/votes/votacoesVotos-{year}.csv"
years = range(2003, 2025)

# Load and concatenate all datasets
df_votes = pd.concat(
    [pd.read_csv(data_path.format(year=year), delimiter=';', quotechar='"') for year in years],
    ignore_index=True
)
```

```python
df_votes
```

```python
# Filter df_votes based on df_sessions
if 'df_sessions' in locals() and 'id' in df_sessions_unique.columns:
    df_votes_selected = df_votes[df_votes['idVotacao'].isin(df_sessions_unique['id'])].copy()
    
    # Merge 'author' column from df_sessions
    df_votes_selected = df_votes_selected.merge(df_sessions_unique[['id', 'author']], 
                                                left_on='idVotacao', right_on='id', how='left')
    df_votes_selected.drop(columns=['id'], inplace=True)
```

```python
df_votes_selected
```

```python
# Ensure the date column is in datetime format
df_votes_selected['dataHoraVoto'] = pd.to_datetime(df_votes_selected['dataHoraVoto'])

# Sort by voting date to ensure chronological processing
df_votes_selected = df_votes_selected.sort_values(by='dataHoraVoto')

# Dictionary to store past votes count per author
author_stats = {}

# List to store results
popularity_records = []

# Iterate through each row in chronological order
for index, row in df_votes_selected.iterrows():
    author = row['author']
    vote = row['voto']
    session_date = row['dataHoraVoto'].date()  # Only consider the date part
    session_id = row['idVotacao']

    # Get past statistics before this voting session (only consider previous days)
    past_yes = 0
    past_total = 0
    for past_date, stats in author_stats.get(author, {}).items():
        if past_date < session_date:  # Only consider votes from previous days
            past_yes += stats['yes_votes']
            past_total += stats['total_votes']

    # Compute popularity before this vote (avoid division by zero)
    popularity = past_yes / past_total if past_total > 0 else 0

    # Store result
    popularity_records.append({'author': author, 'popularity': popularity, 'date': session_date, 'idVotacao': session_id})

    # Update author statistics after this vote
    if author not in author_stats:
        author_stats[author] = {}
    
    if session_date not in author_stats[author]:
        author_stats[author][session_date] = {'yes_votes': 0, 'total_votes': 0}
    
    author_stats[author][session_date]['total_votes'] += 1
    if vote.lower() == 'sim':  # Adjusted for Portuguese dataset
        author_stats[author][session_date]['yes_votes'] += 1

# Create dataframe with computed popularity
df_author_popularity = pd.DataFrame(popularity_records)
```

```python
# Drop duplicate voting session IDs
df_author_popularity = df_author_popularity.drop_duplicates(subset=['idVotacao'])
```
