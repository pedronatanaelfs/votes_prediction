```python
import networkx as nx

# Load the graph from the gpickle file
import pickle

graph_path = "../../data/graphs_with_communities/graph_legislature_57_with_communities.gpickle"
with open(graph_path, "rb") as f:
    G = pickle.load(f)

# Save the graph in GML format
gml_path = "../../data/graphs_with_communities/graph_legislature_57_with_communities.gml"
nx.write_gml(G, gml_path)

print(f"Graph saved to {gml_path} in GML format.")
```
