```python
import pandas as pd
from pathlib import Path
```

```python
# Caminho base dos dados
DATA = Path("../data")

# Função para carregar todos os CSVs de um prefixo
def load_csvs(prefix):
    files = sorted(DATA.glob(f"{prefix}-*.csv"))
    frames = []
    for file in files:
        df = pd.read_csv(file, sep=";", quotechar='"', low_memory=False)
        df["ano"] = int(file.stem.split("-")[-1])  # salva o ano como coluna
        frames.append(df)
    return pd.concat(frames, ignore_index=True)
```

```python
votos = load_csvs("voting/votes/votacoesVotos")

votos
```

```python
# Exemplo: dropando as colunas 'deputado_uri', 'deputado_uriPartido' e 'deputado_urlFoto'
colunas_para_dropar = ['uriVotacao', 'deputado_uri', 'deputado_nome', 'deputado_uriPartido', 'deputado_urlFoto']
votos = votos.drop(columns=colunas_para_dropar)
votos
```

```python
votos.to_csv("votos_agg.csv", index=False, sep=";")
```
