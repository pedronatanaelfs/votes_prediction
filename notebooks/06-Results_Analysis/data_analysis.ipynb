```python
import pandas as pd
```

```python
df = pd.read_csv('votacoes_network.csv')
```

```python
df.head()
```

```python
# Análise da quantidade de 0 e 1 na coluna "aprovacao" para cada valor das colunas especificadas

# Lista de colunas para análise
columns_to_analyze = ["orientacao_GOV", "classified_llm", "legislatura", "autor_community_id", "tema"]

for col in columns_to_analyze:
    print(f"\nQuantidade de 0 e 1 na coluna 'aprovacao' para cada valor de '{col}':")
    print(df.groupby(col)["aprovacao"].value_counts().unstack(fill_value=0))
```

```python
# Para cada orientação do governo, mostrar a contagem de 0 e 1 em 'aprovacao' para cada tema
print("\nContagem de 0 e 1 em 'aprovacao' para cada 'tema' dentro de cada 'orientacao_GOV':")
tema_counts = df.groupby(['orientacao_GOV', 'tema'])['aprovacao'].value_counts().unstack(fill_value=0)
print(tema_counts)

# Para cada orientação do governo, mostrar a contagem de 0 e 1 em 'aprovacao' para cada classified_llm
print("\nContagem de 0 e 1 em 'aprovacao' para cada 'classified_llm' dentro de cada 'orientacao_GOV':")
llm_counts = df.groupby(['orientacao_GOV', 'classified_llm'])['aprovacao'].value_counts().unstack(fill_value=0)
print(llm_counts)
```
