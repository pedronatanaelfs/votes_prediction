```python
import nltk
import pandas as pd
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Download Portuguese stopwords (only needed once)
nltk.download('stopwords')

# Load stopwords for Portuguese
portuguese_stopwords = stopwords.words('portuguese')

# Example dataset (replace with your own data)
texts = [
    "A votação foi aprovada com maioria absoluta.",
    "O projeto de lei foi rejeitado por ampla margem.",
    "O deputado X propôs uma emenda ao texto original.",
    "A sessão foi encerrada após intenso debate.",
    "Houve consenso na aprovação do projeto.",
    "O presidente da câmara vetou a proposta do deputado.",
    "Os parlamentares discutiram alterações na lei tributária.",
    "A emenda foi retirada da pauta após pressão política.",
    "Houve impasse na votação e o projeto foi adiado.",
]

# Convert text to TF-IDF features
vectorizer = TfidfVectorizer(stop_words=portuguese_stopwords, max_features=500)
X = vectorizer.fit_transform(texts)

# Finding the optimal number of clusters using silhouette score
best_k, best_score = None, -1
for k in range(2, 6):  # Try different values for K
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X)
    score = silhouette_score(X, labels)
    if score > best_score:
        best_k, best_score = k, score

# Fit K-Means with the best number of clusters
kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

# Show results
df = pd.DataFrame({'text': texts, 'cluster': labels})
print("\nClustered Texts:")
print(df.sort_values(by="cluster"))
```
